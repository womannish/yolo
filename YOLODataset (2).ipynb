```python
from ultralytics import YOLO
import os

DATA_YAML = "/content/data.yaml"
RESULTS_DIR = "/content/drive/MyDrive/Newgen/results"

os.makedirs(RESULTS_DIR, exist_ok=True)

print("Dataset:", DATA_YAML)
print("Results:", RESULTS_DIR)
```

```python
model_v5 = YOLO("yolov5s.pt")
```

```python
results_v5 = model_v5.train(
    data=DATA_YAML,
    epochs=10,
    imgsz=640,
    batch=16,
    device=0,
    project=RESULTS_DIR,
    name="yolov5s_baseline"
)
```

```python
model_26 = YOLO("yolo26n.pt")
model_26.export(format="engine", device=0)
```

```python
results_26 = model_26.train(
    data=DATA_YAML,
    epochs=10,
    imgsz=640,
    batch=16,
    device=0,
    project=RESULTS_DIR,
    name="yolo26n"
)
```

```python
best_v5 = YOLO(
    "/content/drive/MyDrive/Newgen/results/yolov5s_baseline-4/weights/best.pt"
)

metrics_v5 = best_v5.val(
    data=DATA_YAML,
    imgsz=640,
    device=0
)

print("YOLOv5 Results")
print("mAP50:", metrics_v5.box.map50)
print("mAP50-95:", metrics_v5.box.map)
```

```python
best_26 = YOLO(
    "/content/drive/MyDrive/Newgen/results/yolo26n/weights/best.pt"
)

metrics_26 = best_26.val(
    data=DATA_YAML,
    imgsz=640,
    device=0
)

print("YOLO26 Results")
print("mAP50:", metrics_26.box.map50)
print("mAP50-95:", metrics_26.box.map)
```

```python
print("=" * 50)
print("MODEL COMPARISON")
print("=" * 50)

print(f"YOLOv5s  mAP@0.5:       {metrics_v5.box.map50:.4f}")
print(f"YOLOv5s  mAP@0.5:0.95:  {metrics_v5.box.map:.4f}")

print()

print(f"YOLO26n  mAP@0.5:       {metrics_26.box.map50:.4f}")
print(f"YOLO26n  mAP@0.5:0.95:  {metrics_26.box.map:.4f}")
```

```python
import glob
import time
import numpy as np

test_images = glob.glob(
    "/content/drive/MyDrive/Newgen/Dataset/vehicle dataset/valid/images/*"
)

# Use 100 images for a quick benchmark
test_images = test_images[:900]

print("Images used:", len(test_images))
```

```python
start = time.time()

for img in test_images:
    best_v5.predict(
        source=img,
        imgsz=640,
        device=0,
        verbose=False
    )

v5_total = time.time() - start
v5_avg = (v5_total / len(test_images)) * 1000

print(f"YOLOv5s average inference time: {v5_avg:.2f} ms/image")
```

```python
start = time.time()

for img in test_images:
    best_26.predict(
        source=img,
        imgsz=640,
        device=0,
        verbose=False
    )

v26_total = time.time() - start
v26_avg = (v26_total / len(test_images)) * 1000

print(f"YOLO26n average inference time: {v26_avg:.2f} ms/image")
```

```python
print("\n" + "=" * 60)
print("FINAL COMPARISON")
print("=" * 60)

print(f"{'Metric':<25} {'YOLOv5s':>15} {'YOLO26n':>15}")
print("-" * 60)

print(
    f"{'mAP@0.5':<25} "
    f"{metrics_v5.box.map50:>15.4f} "
    f"{metrics_26.box.map50:>15.4f}"
)

print(
    f"{'mAP@0.5:0.95':<25} "
    f"{metrics_v5.box.map:>15.4f} "
    f"{metrics_26.box.map:>15.4f}"
)

print(
    f"{'Inference (ms/image)':<25} "
    f"{v5_avg:>15.2f} "
    f"{v26_avg:>15.2f}"
)
```
